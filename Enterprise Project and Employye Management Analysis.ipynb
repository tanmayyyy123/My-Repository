```python
# Import the pandas library for data manipulation and analysis
import pandas as pd
```

```python
# Import the numpy library for numerical operations
import numpy as np
```

**Task 1: Create dataframes and save as .csv files**

```python
# Creating 'Project' DataFrame using 'pd.DataFrame()' function
project_data = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005', 'A002', 'A005', 'A003', 'A001', 'A003', 'A001', 'A004', 'A004', 'A005'],
    'Project': ['Project 1', 'Project 2', 'Project 3', 'Project 4', 'Project 5', 'Project 6', 'Project 7', 'Project 8', 'Project 9', 'Project 10', 'Project 11', 'Project 12', 'Project 13', 'Project 14'],
    'Cost': [1002000, 2000000, 4500000, 5500000, np.nan, 680000, 400000, 350000, np.nan, 300000, 2000000, 1000000, 3000000, 200000],
    'Status': ['Finished', 'Ongoing', 'Finished', 'Ongoing', 'Finished', 'Failed', 'Finished', 'Failed', 'Ongoing', 'Finished', 'Failed', 'Ongoing', 'Finished', 'Finished']
}
df_project = pd.DataFrame(project_data)
```

```python
# Print the new "df_project" dataframe
print(df_project)
```

```python
# Creating 'Employee' DataFrame using 'pd.DataFrame()' function
employee_data = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],
    'Name': ['John Alter', 'Alice Luxumberg', 'Tom Sabestine', 'Nina Adgra', 'Amy Johny'],
    'Gender': ['M', 'F', 'M', 'F', 'F'],
    'City': ['Paris', 'London', 'Berlin', 'Newyork', 'Madrid'],
    'Age': [25, 27, 29, 31, 30]
}
df_employee = pd.DataFrame(employee_data)
```

```python
# Print the new "df_emplpoyee" dataframe
print(df_employee)
```

```python
# Creating 'Seniority Level' DataFrame using 'pd.DataFrame()' function
seniority_data = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],
    'Designation Level': [2, 2, 3, 2, 3]
}
df_seniority = pd.DataFrame(seniority_data)
```

```python
# Print the new "df_senority" dataframe
print(df_seniority)
```

```python
# Convert dictionary into a Pandas DataFrame using 'pd.DataFrame()' function
df_project = pd.DataFrame(project_data)
df_employee = pd.DataFrame(employee_data)
df_seniority = pd.DataFrame(seniority_data)
```

```python
# Save dataframes to CSV files using '.csv' and 'index' functions
df_project.to_csv('project.csv', index=False)
df_employee.to_csv('employee.csv', index=False)
df_seniority.to_csv('seniority.csv', index=False)
```

```python
# Load the dataframes back from the saved CSV files using 'pd.read_csv()' function for subsequent tasks
df_project = pd.read_csv('project.csv')
df_employee = pd.read_csv('employee.csv')
df_seniority = pd.read_csv('seniority.csv')
```

**Task 2: Compute missing values in 'Cost' using running average via a For loop**

```python
#Print the current 'Project' dataframe
print(df_project)
```

```python
# Calculate missing 'Cost' using 'range()', 'len(Project)', 'pd.insull()', 'Project.loc()', '.mean()' functions
# Iterate through the rows of the Project dataframe to locate missing values
for i in range(len(df_project)):
    # Check if the current value is not missing
    if pd.isnull(df_project.loc[i, 'Cost']):
        # Calculate the running average
        running_avg = df_project.loc[:i, 'Cost'].mean()
        # Replace all NaN (missing) values in the Cost column with the calculated running average
        df_project.loc[i, 'Cost'] = running_avg
```

```python
# Print the new 'Project' dataframe with missing 'Cost' values
print(df_project)
```

**Task 3: Split the 'Name' column into 'First Name' and 'LastName', and remove older 'Name'**

```python
# Read the Employee DataFrame from the saved CSV file.
df_employee = pd.read_csv("employee.csv")
```

```python
#Print the current 'Employee' dataframe
print(df_employee)
```

```python
# Split the 'Name' column by space into two new columns using '.str.split()' function
df_employee[['First Name', 'LastName']] = df_employee['Name'].str.split(' ', n=1, expand=True)
```

```python
# Drop/ Remove the original 'Name' column from the employee dataframe using '.drop()' function
df_employee = df_employee.drop(columns=['Name'])
```

```python
# Print the new 'Employee' dataframe with 'First' and 'Last' name
print(df_employee)
```

**Task 4: Join all three dataframes into a single dataframe named 'Final'**

```python
# Merge all three (Project + Employee + Seniority Level) dataframes into one 'Final' (final) DataFrame using 'pd.merge()' function
df_final = df_project.merge(df_employee, on="ID").merge(df_seniority, on="ID")
```

```python
# Print the new joined/ merged 'Final' dataframe
print(df_final)
```

**Task 5: Add a 'Bonus' column and 5% of project cost for finished projects**

```python
# Print the current 'Final' dataframe
print(df_final)
```

```python
# Calculate the bonus using 'np.where()' function; Bonus = 5% of Cost if project is 'Finished', else 0
df_final["Bonus"] = np.where(df_final["Status"]=="Finished", df_final["Cost"]*0.05, 0)
```

```python
# Print the new 'Final' dataframe with 5% bonus on 'Finished' projects
print(df_final)
```

**Task 6: Demote designation level by 1 for failed projects, delete if above 4**

```python
# Print the currnet 'Final' dataframe (Project 6, 8, and 11 are Failed)
print(df_final)
```

```python
# Identify employees who have 'Failed' projects
failed_projects = df_final["Status"] == "Failed"
```

```python
# Demote employees by increasing their designation number by one (highest is 1, lowest is 4) using '.loc[]' function
df_final.loc[failed_projects, "Designation Level"] = df_final.loc[
    failed_projects, "Designation Level"
] + 1
```

```python
# Remove records where designation level is greater than four
df_final = df_final[df_final["Designation Level"] <= 4]
```

```python
# Reset the DataFrame index
df_final = df_final.reset_index(drop=True)
```

```python
# Print the new 'Final' dataframe with updated 'Designation Level'
print(df_final)
```

**Task 7: Add 'Mr.' or 'Mrs.' to 'First Name' based on Gender, then drop 'Gender'**

```python
#Print current 'Final' dataframe which has 'Gender' coloumn
print(df_final)
```

```python
# Add Mr. or Mrs. prefix based on 'Gender' using 'np.where()' function
df_final["First Name"] = np.where(df_final["Gender"]=="M","Mr. "+df_final["First Name"],"Mrs. "+df_final["First Name"])
```

```python
# Drop 'Gender' column using '.drop()' function
df_final.drop(columns=["Gender"], inplace=True)
```

```python
# Print the new 'Final' dataframe with updated 'First Name'
print(df_final)
```

**Task 8: Promote designation level by 1 for employees whose age > 29 using "IF' condition**

```python
#Print the 'Final' dataframe to check the employees with age >29
print(df_final)
```

```python
# Decrease designation level by 1 (promotion) if Age > 29 using '.loc()' fucntion
df_final.loc[df_final["Age"]>29,"Designation Level"] -= 1
```

**OR**

```python
# Decrease designation level by 1 (promotion) if Age > 29 using 'np.where()' function
df_final['Designation Level'] = np.where(
    df_final['Age'] > 29,
    df_final['Designation Level'] - 1,
    df_final['Designation Level']
)
```

```python
# Print the new 'Final' dataframe with updated designation
print(df_final)
```

**Task 9: Calculate total project cost for each employee and save in 'TotalProjCost' dataframe**

```python
# Creat a new 'Total Poject Cost' dataframe and Group the 'Final' dataframe by 'ID' and 'First Name', summing up the project 'Cost' and reset the index using '.groupby()', '.sum()', and 'reset_index()' function
df_TotalProjCost = df_final.groupby(["ID","First Name"])["Cost"].sum().reset_index()
```

```python
# Rename the 'Cost' column to 'Total cost' to match requirements using '.rename()' function
df_TotalProjCost.rename(columns={"Cost":"Total Cost"}, inplace=True)
```

```python
# Print the new "TotalProjCost" dataframe with updated title
print(df_TotalProjCost)
```

**Task 10: Print employee details whose city name contains the letter 'o'**

```python
# Print the current 'Final' dataframe
print(df_final)
```

```python
# Filter employees whose City contains letter 'o' using '.str.contains()' function
df_employee_city_o = df_final[df_final["City"].str.contains("o", case=False)]
```

```python
# Print the final filtered employee details
print(df_employee_city_o)
```
